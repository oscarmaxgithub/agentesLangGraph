```python
#from langchain_openai import ChatOpenAI
from langchain_openrouter import ChatOpenRouter


# gpt4o_llm= ChatOpenAI(model="gpt-4o", temperature=0)
model = ChatOpenRouter(
    model="openrouter/free",
    temperature=0,
)

response = model.invoke("Hola como estas?")
response.text
```

```python
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage
from langchain_core.runnables import history


msg_1 = SystemMessage(content="Tu eres un asistente de ayuda que habla en español")
msg_2 = HumanMessage(content="Me llamo Oscar")
msg_3 = AIMessage(content="Hola Oscar, como estas?")
msg_4 = HumanMessage(content="Como me llamo?")
history = [msg_1,msg_2,msg_3,msg_4]
response= model.invoke(history)
response.text
```
