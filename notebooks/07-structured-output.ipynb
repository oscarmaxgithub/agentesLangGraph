```python
from langchain_openrouter import ChatOpenRouter

llm = ChatOpenRouter(
    model="openrouter/free",
    temperature=1,
)
response = llm.invoke("Hola soy Oscar Diaz y mi telefono es  +51 975638363")
response.text
```

```python
from email import message
from pydantic import BaseModel, Field


class ContactInfo(BaseModel):
    """ Contact information for a person """
    name: str =  Field(description= "The name of the person")
    email: str =  Field(description= "The email of the person")
    phone: str =  Field(description= "The phone number of the person")
    tone: int = Field(description="The tone of the person", ge=0, le=100)
    age: int = Field(description="The tone of the person")
    sentiment: str =  Field(description= "The sentiment convertation of the person")

llm_with_strutured_output = llm.with_structured_output(schema=ContactInfo)
messages = [
    ("system", "You are a helpfull assistant thath cant extract contact informacion form a given a conversation."),
    # ("user",response.text)
    ("user","Hola soy Oscar Diaz y mi telefono es  +51 975638363")
]
response = llm_with_strutured_output.invoke(messages)
response
```
